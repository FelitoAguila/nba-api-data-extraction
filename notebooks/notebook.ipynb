```python
from nba_api.stats.static import players
import pandas as pd

# get_players returns a list of dictionaries, each representing a player.
nba_players = players.get_players()
print(f"Number of players fetched: {len(nba_players)}")

nba_players_df = pd.DataFrame(nba_players)
```

```python
tyrese_maxey = nba_players_df[nba_players_df['full_name'] == 'Tyrese Maxey']

# Maxey Career Stats
career = playercareerstats.PlayerCareerStats(player_id=tyrese_maxey['id'].iloc[0])
maxey_career_df = career.get_data_frames()[0]
```
